```python
"""
MEDIUM RESOLUTION TRAINING (960px)
Faster than 1280px but still much better than 640px
Training time: ~2.5 hours instead of 4-5 hours
"""

print("="*70)
print("MEDIUM RESOLUTION WALL DETECTION (960px)")
print("="*70)
print("\n⚡ COMPROMISE: Faster training with good improvement")
print("  640px → 960px = 2.25x more pixels per object")
print("  Training time: ~2.5 hours (vs 4-5 hours for 1280px)\n")

import subprocess
import sys
import os

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "ultralytics"])

from ultralytics import YOLO
import yaml

# Configuration
DATASET_ROOT = "/content/copy-of-the-dataset-1"
IMAGE_SIZE = 960
MODEL_SIZE = 's'
EPOCHS = 120
BATCH_SIZE = 8

print("⚙️ SETTINGS:")
print(f"  Resolution: {IMAGE_SIZE}px (1.5x your current)")
print(f"  Model: YOLOv8{MODEL_SIZE}")
print(f"  Epochs: {EPOCHS}")
print(f"  Expected wall improvement: +20-35%\n")

# Setup data
if os.path.exists(os.path.join(DATASET_ROOT, 'train_enhanced')):
    train_path = os.path.join(DATASET_ROOT, 'train_enhanced/images')
else:
    train_path = os.path.join(DATASET_ROOT, 'train/images')

data_config = {
    'train': train_path,
    'val': os.path.join(DATASET_ROOT, 'valid/images'),
    'test': os.path.join(DATASET_ROOT, 'test/images'),
    'nc': 3,
    'names': ['door', 'wall', 'window']
}

config_path = os.path.join(DATASET_ROOT, 'data_960.yaml')
with open(config_path, 'w') as f:
    yaml.dump(data_config, f)

print("🏋️ Starting training...\n")

model = YOLO(f'yolov8{MODEL_SIZE}.pt')

results = model.train(
    data=config_path,
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    name='wall_960px',
    patience=25,

    # Optimized for small objects
    box=9.0,
    cls=0.6,
    dfl=2.0,

    optimizer='AdamW',
    lr0=0.001,
    lrf=0.01,

    # Moderate augmentation
    hsv_h=0.015,
    hsv_s=0.5,
    hsv_v=0.3,
    degrees=10.0,
    translate=0.1,
    scale=0.25,
    fliplr=0.5,
    mosaic=0.4,

    device=0,
    workers=8,
    verbose=True,
    plots=True,
    save=True,
)

print("\n✓ Training complete!\n")

# Evaluate
metrics = model.val()

print("\n" + "="*70)
print("960px MODEL RESULTS")
print("="*70)

class_names = ['door', 'wall', 'window']
for i, name in enumerate(class_names):
    p = metrics.box.p[i]
    r = metrics.box.r[i]
    f1 = 2 * p * r / (p + r + 1e-6)
    print(f"\n{name.upper()}:")
    print(f"  Precision: {p:.4f}")
    print(f"  Recall:    {r:.4f}")
    print(f"  F1:        {f1:.4f}")
    print(f"  mAP50:     {metrics.box.ap50[i]:.4f}")

print("\n" + "="*70)

# Save
import shutil
try:
    shutil.copy2('runs/detect/wall_960px/weights/best.pt',
                 '/content/best_wall_960px.pt')
    print("✓ Model saved: /content/best_wall_960px.pt")
    print("\nDownload with:")
    print("from google.colab import files")
    print("files.download('/content/best_wall_960px.pt')")
except:
    pass

print("="*70)
```
